```python
import numpy as np

x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print(x_train.shape)
print(x_train[1])
print(y_train[1])
```

```python
import matplotlib.pyplot as plt

x_train = np.array([1.0, 2.0, 3.5,4.0])
y_train = np.array([300.0, 500.0, 700.0, 900.0])

m = x_train.shape[0]
print(x_train.shape)
print(m)
print(x_train[2])
plt.scatter(x_train, y_train, marker ="x", c="r")
plt.xlabel("Size")
plt.ylabel("Price")
plt.show()
```

## What I learned (2026-09-14)
- shape returns a tuple; (4,) means 1d array with 4 elements
- index = position, value = content.x_train[2] reads position 2.
- x_train[0] and x_train.shape[0] index different objects.
- scatter pairs the two arrays by the same indexing, not by sorted order.
