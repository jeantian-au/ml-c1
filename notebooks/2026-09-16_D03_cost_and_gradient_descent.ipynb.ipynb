Cost Functions:
- What the cost function measures:
  The cost functions measure how far the model predictions are to the actual target values across a training set.

- Why squared error, not sum or mean:
  Because the predicted value - actual value might be a negative number, so the squared will guarantee the result would always be postive.
The cost function squares the loss ensures that the error surface is convex like a soup bowl. It will always have a minimum that can be reached by following the gradient in all dimensions. 

- Symmetry of J(w):
  When b is fixed, the figure will be convex parabola with a lowest point as the J(w)'s minimum point, and it's symmetric.

- Where the minimum sits:
  the position on the w axis is set by the data, not by zero. 

- Contour plots:
  A contour plot shows combinations of parameters that produce the same cost. Each contour line represents a constant cost value.
  Moving along a contour line leaves cost unchanged,so the steepest descent direction is perpendicular to the contour line.

- convex vs. local minima:
  A local minimum has the lowest cost within a nearby region, while a global minimum has the lowest cost over the entire domain.
  For a convex cost function, every local minimum is also a global minimum. The squared-error cost function for linear regression is convex.
